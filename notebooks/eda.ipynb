```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("../data/Dataset.csv")
```

```python
print(data.sample(3))
```

```python
print(data.info())
print("Shape of Data:",data.shape)
data.columns
```

```python
data["TotalCharges"] = pd.to_numeric(data["TotalCharges"],errors="coerce")
```

```python
print(data.describe())
data["TotalCharges"] = data["TotalCharges"].fillna(data["TotalCharges"].mean())
print(data.isnull().sum())
```

```python
fig,axes = plt.subplots(5,3,figsize=(18,12))
sns.countplot(x=data["gender"],ax=axes[0,0])
sns.countplot(x=data["Dependents"],ax=axes[0,1])
sns.countplot(x=data["InternetService"],ax=axes[0,2])
sns.countplot(x=data["Partner"],ax=axes[1,0])
sns.countplot(x=data["Contract"],ax=axes[1,1])
sns.countplot(x=data["DeviceProtection"],ax=axes[1,2])
sns.countplot(x=data["SeniorCitizen"],ax=axes[2,0])
sns.countplot(x=data["PaperlessBilling"],ax=axes[2,1])
sns.countplot(x=data["OnlineSecurity"],ax=axes[2,2])
sns.countplot(x=data["PhoneService"],ax=axes[3,0])
sns.countplot(x=data["OnlineBackup"],ax=axes[3,1])
sns.countplot(x=data["TechSupport"],ax=axes[3,2])
sns.countplot(x=data["PaymentMethod"],ax=axes[4,0])
sns.countplot(x=data["StreamingTV"],ax=axes[4,1])
sns.countplot(x=data["Churn"],ax=axes[4,2])
axes[0,0].set_title("Gender Column Distribution")
axes[0,1].set_title("Dependent Column Distribution")
axes[0,2].set_title("InternetService Column Distribution")
axes[1,0].set_title("Partner Column Distribution")
axes[1,1].set_title("Contract Column Distribution")
axes[1,2].set_title("DeviceProtection Column Distribution")
axes[2,0].set_title("SeniorCitizen Column Distribution")
axes[2,1].set_title("PaperlessBilling Column Distribution")
axes[2,2].set_title("OnlineSecurity Column Distribution")
axes[3,0].set_title("PhoneService Column Distribution")
axes[3,1].set_title("OnlineBackup Column Distribution")
axes[3,2].set_title("TechSupport Column Distribution")
axes[4,0].set_title("PaymentMethod Column Distribution")
axes[4,1].set_title("StreamingTV Column Distribution")
axes[4,2].set_title("Churn Column Distribution")
plt.tight_layout()
plt.show()
```

```python
fig,axes = plt.subplots(3,1,figsize=(10,15))
sns.histplot(data["tenure"],kde=True,ax=axes[0])
axes[0].set_title("Tenure Column Distribution")
sns.histplot(data["MonthlyCharges"],kde=True,ax=axes[1])
axes[1].set_title("MonthlyCharges Column Distribution")
sns.histplot(data["TotalCharges"],kde=True,ax=axes[2])
axes[2].set_title("TotalCharges Column Distribution")
```

```python
fig,axes = plt.subplots(3,1,figsize=(10,15))
sns.boxplot(x=data["tenure"],ax=axes[0])
axes[0].set_title("Tenure Column Distribution")
sns.boxplot(x=data["MonthlyCharges"],ax=axes[1])
axes[1].set_title("Monthly Charges Column Distribution")
sns.boxplot(x=data["TotalCharges"],ax=axes[2])
axes[2].set_title("Total Charges Column Distribution")
```

```python
fig,axes = plt.subplots(3,2,figsize=(14,14))
sns.boxplot(x="Churn", y="TotalCharges", data=data,ax=axes[0,0])
axes[0,0].set_title("Customers with respect to total charges")
sns.boxplot(x="Churn", y="MonthlyCharges", data=data,ax=axes[0,1])
axes[0,1].set_title("Customers with respect to monthly charges")
sns.boxplot(x="Churn", y="tenure", data=data,ax=axes[1,0])
axes[1,0].set_title("Customers with respect to tenure")
sns.countplot(x="Churn", hue="Contract", data=data,ax=axes[1,1])
axes[1,1].set_title("Customers with respect to Contract")
sns.countplot(x="Churn", hue="PaymentMethod", data=data,ax=axes[2,0])
axes[2,0].set_title("Customers with respect to PaymentMethod")
sns.countplot(x="Churn",hue="InternetService", data=data,ax=axes[2,1])
axes[2,1].set_title("Customers with respect to InternetService")
plt.tight_layout()
plt.show()
```

```python
fig,axes = plt.subplots(4,2,figsize=(20,14))
sns.countplot(x="Churn", hue="gender", data=data,ax=axes[0,0])
axes[0,0].set_title("Customers churn rate with respect to gender")
sns.countplot(x="Churn", hue="Partner", data=data,ax=axes[0,1])
axes[0,1].set_title("Customers churn rate with respect to partner")
sns.countplot(x="Churn", hue="Dependents", data=data,ax=axes[1,0])
axes[1,0].set_title("Customers churn rate with respect to Dependents")
sns.countplot(x="Churn", hue="TechSupport", data=data,ax=axes[1,1])
axes[1,1].set_title("Customers churn rate with respect to TechSupport")
sns.countplot(x="Churn", hue="SeniorCitizen", data=data,ax=axes[2,0])
axes[2,0].set_title("Customers churn rate with respect to SeniorCitizen")
sns.countplot(x="Churn", hue="MultipleLines", data=data,ax=axes[2,1])
axes[2,1].set_title("Customers churn rate with respect to MultipleLines")
sns.countplot(x="Churn", hue="StreamingTV", data=data,ax=axes[3,0])
axes[3,0].set_title("Customers churn rate with respect to StreamingTV")
sns.countplot(x="Churn", hue="StreamingMovies", data=data,ax=axes[3,1])
axes[3,1].set_title("Customers churn rate with respect to StreamingMovies")
plt.tight_layout()
plt.show()
```
