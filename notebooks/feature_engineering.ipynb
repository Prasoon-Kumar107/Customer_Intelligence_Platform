```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../data/Dataset.csv")
df.sample(3)
```

```python
new_df = df.copy()
print(new_df.sample(3))
df.columns
```

```python
new_df = new_df.drop(["customerID","gender"],axis=1)
```

```python
def familysize(row):
    if row["Partner"] == "Yes" and row["Dependents"] == "No":
        return "has spouse"
    elif row["Partner"] == "No" and row["Dependents"] == "No":
        return "Alone"
    else:
        return "Has family"
new_df["familytype"] = new_df.apply(familysize,axis=1)
df["familytype"] = df.apply(familysize,axis=1)
```

```python
new_df = new_df.drop(["Partner","Dependents"],axis=1)
```

```python
new_df["SecurityScore"] = (
    (new_df["OnlineSecurity"] == "Yes").astype(int)
    + (new_df["OnlineBackup"] == "Yes").astype(int)
    + (new_df["DeviceProtection"] == "Yes").astype(int)
    + (new_df["TechSupport"] == "Yes").astype(int)
)
df["SecurityScore"] = (
    (df["OnlineSecurity"] == "Yes").astype(int)
    +(df["OnlineBackup"] == "Yes").astype(int)
    +(df["DeviceProtection"] == "Yes").astype(int)
    +(df["TechSupport"] == "Yes").astype(int)
)
new_df = new_df.drop(["OnlineSecurity","OnlineBackup","DeviceProtection","TechSupport"],axis=1)
```

```python
new_df = new_df.drop(["PaperlessBilling"],axis=1)
print(new_df.sample(1))
```

```python
new_df.sample(5)
```

```python
new_df["EntertainmentScore"] = (
    (new_df["StreamingTV"] == "Yes").astype(int)
    +(new_df["StreamingMovies"] == "Yes").astype(int)
)
df["EntertainmentScore"] = (
    (df["StreamingTV"] == "Yes").astype(int)
    +(df["StreamingMovies"] == "Yes").astype(int)
)
new_df = new_df.drop(["StreamingTV","StreamingMovies"],axis=1)
```

```python
new_df.sample(3)
```

```python
df.sample(3)
```

```python
def duration(row):
    if row["tenure"] > 25:
        return "Old"
    elif row["tenure"] < 10:
        return "New"
    else:
        return "Moderate"
new_df["Customertype"] = new_df.apply(duration,axis=1)
df["Customertype"] = df.apply(duration,axis=1)
new_df = new_df.drop(["tenure"],axis=1)
```

```python
new_df.sample(3)
```

```python
df.sample(3)
```

```python
def risk(row):
    if (row["Contract"] == "Month-to-month" and row["PaymentMethod"] == "Electronic check") and (row["InternetService"] == "Fiber optic" and row["TechSupport"] == "No"):
        return "High risk"
    elif (row["Contract"] == "Month-to-month" and row["PaymentMethod"] == "Mailed check") and (row["InternetService"] == "DSL" and row["SeniorCitizen"] == 0):
        return "Medium risk"
    else:
        return "Low risk"
df["RiskType"] = df.apply(risk,axis=1)
```

```python
print(df.sample(3))
df.columns
```

```python
df = df.drop(["StreamingTV","StreamingMovies","gender","MultipleLines"],axis=1)
```

```python
print(df.sample(3))
df.columns
```

```python
df["Churn_binary"] = df["Churn"].map({
    "Yes": 1,
    "No": 0
})
risk_churn = (
    df.groupby("RiskType")["Churn_binary"]
      .mean()
      .mul(100)
      .round(2)
)
print(risk_churn)
df["RiskType"].value_counts()
df.groupby("RiskType")["Churn_binary"].agg(
    ["count", "mean"]
)
df.groupby("RiskType")["Churn_binary"].agg(
    ["count", "mean"]
)
```
